#### Lab 1: Recursive Function

```python
# A recursive function to calculate factorial
def factorial(n):
    # Base condition
    if n == 0:
        return 1
    else:
        return n * factorial(n-1) # use of function within the function

# Calculating the factorial of 5
result = factorial(5)
print(result) # Output: 120
```

```python
#Sum of digit by using recursive function
def digit(n):
    if n<10:
        return n
    else:
        return n%10+digit(n//10)

n=int(input('n='))
s=digit(n)
print(f'Sum of digit of {n}=',s)
```

```python
#From a given list find the maximum and minimum number without using inbuilt function and use recursive function
def max_f(l):
    if len(l)==1:
        return l[0]
    else:
        return max(l[0],max_f(l[1:]))
nums=[2,8,5,9,11,16,7]
max_f(nums)
```

```python
#Power with recursive function n to the power k
def power(n,k):
    if k==0:
        return 1
    else:
        return n*power(n,k-1)
power(3,3)
```

```python
#Find the length of a string by using recursive function
def find_l(s):
    if s=='':
        return 0
    else:
        return 1 + find_l(s[1:])
s='Badriah'
find_l(s)
```

```python
#Calculate the arithmetic sum from 1 to n
def sum_till(n):
    if n==1:
        return 1
    else:
        return n+sum_till(n-1)
sum_till(5) # 1+2+3+4+5 = 15
```

```python
#Create a recursive Python function to find the nth Fibonacci number.
def find_fib(n):
    if n<=1:
        return 1
    else:
        return find_fib(n-1)+find_fib(n-2)
find_fib(6)
```

```python
# Write a recursive function that will count a character will come how many times with in string
def count_c(s,c):
    if s=='':
        return 0
    elif s[0]==c:
        return 1+count_c(s[1:],c)
    else:
        return count_c(s[1:],c)
s='Badriah'
count_c(s,'a')
```

#### Lab 4: Data Reshaping, Filtering, & Handle the Missing Values.

```python
#Import packages
import pandas as pd #for dataframes (like Excel)
import numpy as np #Numerical Operations

#Assign data
#data is a Dictionary of Lists
data = {'NAME':['A','B','C','D','E','F','G'],
        'AGE':[22,23,21,24,25,26,27],
        'GENDER':['M','F','M','M','F','F','M'],
        'MARKS':[78,70, np.nan ,85,89,82,np.nan]}
# np.nan indicates missing values

#Convert into DataFrame
df=pd.DataFrame(data)
#Display the Data
df
```

```python
#Categorize gender
df['GENDER']=df['GENDER'].map({'M':0,'F':1})
#Mapping means convert one set of values to another.
#'M' is mapped (converted) to 0 and 'F' is mapped to 1
df
```

```python
#Shows the selected column as DataFrame (table)
df.filter(['NAME'])
```

```python
#Shows column with missing value as boolean
df.isnull().any()
```

```python
# Calculate the mean
mean = df['MARKS'].mean()
print(mean)
```

```python
# Replace NaN values with mean
df['MARKS'].fillna(mean)
df['MARKS'].mean()
df.dropna() #Remove rows with any NaNs
```
