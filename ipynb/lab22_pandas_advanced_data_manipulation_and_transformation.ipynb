```python
# Data Inspection & Filtering

# ---- Count categories in a column ----
categories = pd.Categorical(df['species'])
print(categories) # Shows unique species: Setosa, Versicolor, Virginica

# ---- Find maximum value in a column ----
max_width = max(df['sepal_width'])
print(max_width)
```

```python
# Filtering by Maximum Value & Setting Index

# ---- Filter rows where column = max value ----
max_width_rows = df[df['sepal_width'] == max(df['sepal_width'])]
print(max_width_rows)

# ---- Set index temporarily ----
temp_index = df.set_index('sepal_width')
print(temp_index) # Only for display, df itself unchanged

# ---- Set index permanently ----
# df.set_index('class', inplace=True) # Uncomment if 'class' column exists
```

```python
# Reset Index & Drop NA Values

# ---- Reset index ----
df_reset = df.reset_index()
print(df_reset)

# ---- Drop rows with NA values temporarily ----
df_drop_temp = df.dropna()
print(df_drop_temp)

# ---- Drop rows permanently ----
# df.dropna(inplace=True)

# ---- Drop columns with NA values ----
# df.dropna(axis=1, inplace=True)
```

```python
# Grouping Data

# ---- Group by species ----
g = df.groupby('species')

# ---- Sum per group ----
print(g.sum())

# ---- Mean per group ----
print(g.mean())

# ---- Max per group ----
print(g.max())
```
