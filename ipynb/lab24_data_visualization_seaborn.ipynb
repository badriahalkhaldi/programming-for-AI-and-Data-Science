```python
df = sns.load_dataset('iris')
df.head()
```

```python
# Scatter Plot
sns.scatterplot(x=df.sepal_length, y=df.sepal_width, color = 'purple')
```

```python
sns.scatterplot(x='sepal_length', y='petal_length', data=df, style = 'species', hue = 'species')
#Each dot represent a flower
```

```python
# Histogram
sns.displot(df['sepal_length'], color = 'red')
```

```python
# Pair plots show all relationships between multiple numerical variables:
sns.pairplot(df, hue = 'species')
```

```python
# Create tips dataset manually
df = pd.DataFrame({
    'total_bill': [10, 20, 15, 30, 25, 18, 40],
    'tip': [2, 3, 2.5, 5, 4, 3, 6],
    'smoker': ['Yes', 'No', 'No', 'Yes', 'No', 'Yes', 'No'],
    'day': ['Sun', 'Sat', 'Sun', 'Fri', 'Sat', 'Fri', 'Sun'],
    'time': ['Dinner', 'Dinner', 'Lunch', 'Dinner', 'Lunch', 'Dinner', 'Lunch'],
    'size': [2, 3, 2, 4, 3, 2, 5]
})

df.head()
```

```python
# Scatter plot of total bill vs tip:
sns.scatterplot(x=df.total_bill, y=df.tip)
```

```python
# Color by smoker/non-smoker:
# Relational Plot - More advanced & flexible than scatter plot
# hue means color by category: column name must be string
sns.relplot(x=df.total_bill, y=df.tip, data=df, hue='smoker')
```

```python
# Style by size of the group:
sns.relplot(x=df.total_bill, y=df.tip, data=df, style='size')
```

```python
# Combine hue and style (e.g., time and size):
sns.relplot(x=df.total_bill, y=df.tip, data=df, hue='time', style='size')
```

```python
# Categorical Plots - These plots compare numerical data across categories:
sns.catplot(x=df.day, y=df.total_bill, data=df)       # Total bill per day
sns.catplot(x='smoker', y='tip', data=df)            # Tip by smoker status
sns.catplot(x='smoker', y='total_bill', data=df)     # Total bill by smoker
```

A joint plot shows the relationship between two variables plus their distributions.

It combines:

- Scatter plot / regression plot in the middle

- Histograms / KDEs on top and right


kind can be:

'scatter' → normal scatter plot

'reg' → scatter plot with regression line

'hex' → hexbin plot for dense data

'kde' → density plot

```python
sns.jointplot(x='total_bill', y='tip', data=df, kind='scatter')
sns.jointplot(x='total_bill', y='tip', data=df, kind='reg')
```

- Seaborn makes plotting easier and prettier than Matplotlib.

- Scatter plots → relationships between numbers

- Histograms → distributions

- Pair plots → all relationships at once

- relplot → scatter plots with additional features like color (hue) or style

- catplot → compare numbers across categories
