```python
cart = []

while True:
    print("\nShopping Cart Menu")
    print("1. Add an item")
    print("2. View items")
    print("3. Remove an item")
    print("4. Exit")
    choice = input("Choose an option (1-4): ")
    if choice == "1":
        item = input("Enter item name: ")
        cart.append(item)
        print(item, "added to cart.")
    elif choice == "2":
        if len(cart) == 0:
            print("Cart is empty.")
        else:
            print("Items in cart:")
            for item in cart:
                print("-", item)
    elif choice == "3":
        item = input("Enter item name to remove: ")
        if item in cart:
            cart.remove(item)
            print(item, "removed from cart.")
        else:
            print("Item not found in cart.")
    elif choice == "4":
        print("Goodbye!")
        break
    else:
        print("Invalid choice. Try again.")
```
