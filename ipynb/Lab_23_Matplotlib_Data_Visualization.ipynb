### Lab 1: Data Visualization by Matplotlib

```python
#Import Libraries
import numpy as np 
import matplotlib.pyplot as plt
import pandas as pd
```

```python
#Create 200 numbers from 1 to 20
x=np.linspace(1,20,200)
#Apply sine function
y=np.sin(x)
#Line Plot
plt.plot(x,y)
```

```python
plt.xlabel('This is my X data')
plt.ylabel('This is my Y data')
plt.title('This is my title')
#Change line style/color
plt.plot(x,y, '-r') #Solid line 
plt.plot(x,y, '--g') #Dashed line
plt.plot(x,y, color='green', linestyle='dashed') #Dashed line
plt.plot(x,y, ':m') #Dotted line
plt.plot(x,y, '-.y') #Dashdot line
# y - yellow, w - white, c - cyan, k - black
```

```python
#Draw a scatter plot
a=np.random.rand(50) #Make 50 random points
b=np.random.rand(50)
plt.scatter(a,b, c='r', marker='^', s=50)
# c is for color , s is for size
# marker is the shape: o (circle), s (square), 
# + (plus sign), ^ (triangles)
```

```python
# Draw a scatter plot with different size & colors
x=np.random.rand(50)
y=np.random.rand(50)
colours=np.random.rand(50)
size=100*(np.random.rand(50))
plt.scatter(x,y,c=colours,s=size)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('X vs Y')
plt.show()
```

```python
plt.scatter(x,y,c=colours,s=size,alpha=0.5)
#Alpha adds transparency, lower the number more transparent
```

```python
#Custom HTML Colour Code
plt.scatter(x,y,c='#FF33E9',s=size)
```

```python
x=[4,5,6,7,8]
y=[60,17,28,96,10]
colour=['red','blue','yellow','green','black']
plt.scatter(x,y,c=colour,marker='D') #Diamond Shape
```

```python
#Bar Graphs - Compare Categories
#Vertical Bar Graph 
x=['a','b','c','d','e']
y=np.random.rand(5)
plt.bar(x,y)
```

```python
plt.figure(figsize=(10,5)) #Change Figure Size 
plt.barh(x,y) #Horizontal bar graph
plt.show()
```

```python
#Histogram: Shows how often each value appears
data=[1,2,3,4,5,1,1,1,2,2,3,3,4,5,5,6,4]
plt.hist(data)
```

```python
x=np.random.rand(50)
y=np.random.rand(50)
z=np.random.rand(50)
fig=plt.figure()
ax=fig.add_subplot(projection='3d')
ax.scatter(x,y,z)
```
