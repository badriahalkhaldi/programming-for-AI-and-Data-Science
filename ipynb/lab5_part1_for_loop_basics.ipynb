```python
fruits = ["apple","banana","mango","kiwi","grapes"]
find = str(input("enter a fruit to find:"))
found = False
for  fruit in (fruits):
    if find==fruit:
        print(find," is in the list ")
        found = True
        break
if found==False:
    print(find," is not in the list ")
```

```python
numbers = [3,5,8,2,5,9,1,13,7,9]
largest= numbers[0]
for n in numbers:
    if n > largest:
        largest = n
print ("the  largest number is ",  largest)
```

```python
numbers = [3,5,8,2,5,9,1,13,7,9]
largest= numbers[0]
for n in numbers:
    if n + largest:
        largest = n
print(largest/10)
print ("the number is ",  largest)
```

```python
sum= 0
length = len(numbers)
for n in numbers:
    sum = sum + n
average = sum/length
print("the average for the given list of numbers is :",average)
```
