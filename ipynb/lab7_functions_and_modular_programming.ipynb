```python
def calculate(num1, num2, operation):
    if operation == "add":
        return num1 + num2
    elif operation == "subtract":
        return num1 - num2
    elif operation == "multiply":
        return num1 * num2
    elif operation == "divide":
        if num2 == 0:
            return "Error: Cannot divide by zero."
        return num1 / num2
    elif operation == "power":
        return num1 ** num2
    else:
        return "Invalid operation."


print("Available operations: add, subtract, multiply, divide, power")

while True:
    op = input("\nEnter operation (or type 'exit' to quit): ").lower()
    if op == "exit":
        print("Goodbye!")
        break

    n1 = float(input("Enter first number: "))
    n2 = float(input("Enter second number: "))

    result = calculate(n1, n2, op)
    print("Result:", result)
```

```python
def fitness_report(steps):
    total = sum(steps)
    average = total / len(steps)
    highest = max(steps)
    lowest = min(steps)

    print("\n--- Weekly Fitness Report ---")
    print("Total steps:", total)
    print("Average steps per day:", average)
    print("Highest steps:", highest)
    print("Lowest steps:", lowest)

    if average >= 10000:
        print("Excellent! Keep it up!")
    elif 7000 <= average < 10000:
        print("Good! Try to reach 10,000 next time.")
    else:
        print("You need to move more!")


# Example usage:
week_steps = []
print("Enter your steps for 7 days:")

for i in range(7):
    day = int(input(f"Day {i+1} steps: "))
    week_steps.append(day)

fitness_report(week_steps)
```

```python
def add_employee():
    name = input("Enter employee name: ")
    emp_id = input("Enter employee ID: ")
    dept = input("Enter department: ")

    with open("employees.txt", "a") as file:
        file.write(f"{name},{emp_id},{dept}\n")
    print("Employee added successfully!")


def display_employees():
    print("\n--- Employee List ---")
    try:
        with open("employees.txt", "r") as file:
            data = file.readlines()
            if not data:
                print("No employees found.")
            else:
                for line in data:
                    name, emp_id, dept = line.strip().split(",")
                    print(f"Name: {name} | ID: {emp_id} | Dept: {dept}")
    except FileNotFoundError:
        print("No employee file found.")


def search_employee():
    search_name = input("Enter employee name to search: ").lower()

    found = False
    try:
        with open("employees.txt", "r") as file:
            for line in file:
                name, emp_id, dept = line.strip().split(",")
                if search_name == name.lower():
                    print(f"Found → Name: {name}, ID: {emp_id}, Dept: {dept}")
                    found = True
                    break
        if not found:
            print("Employee not found.")
    except FileNotFoundError:
        print("No employee file found.")


while True:
    print("\n--- Employee Management System ---")
    print("1. Add Employee")
    print("2. Display All Employees")
    print("3. Search Employee")
    print("4. Exit")

    choice = input("Choose an option: ")

    if choice == "1":
        add_employee()
    elif choice == "2":
        display_employees()
    elif choice == "3":
        search_employee()
    elif choice == "4":
        print("Goodbye!")
        break
    else:
        print("Invalid choice. Try again.")
```
