```python
# Custome Exeption
class BalanceError(Exception):
    pass
class Customer:
    def __init__(self, name, balance):
        if balance < 0:
            raise BalanceError('Balance has to be non-negative!')
        else:
            self.name = name
            self.balance = balance
try:
    cust = Customer('Larry Torres', -100)
except BalanceError:
    cust = Customer('Larry Torres', 0)
print(cust.name)
print(cust.balance)
```
