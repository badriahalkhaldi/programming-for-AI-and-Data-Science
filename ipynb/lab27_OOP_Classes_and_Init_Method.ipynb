```python
class Customer:
    def __init__(self, name, balance):
        self.name = name
        self.balance = balance
        print("The __init__ method was called")

cust = Customer("Lara de Silva", 1000)

print(cust.name)
print(cust.balance)
```

```python
class Employee:
    def __init__(self, name, salary=0):
        self.name = name
        self.salary = salary

emp1 = Employee('Mashael', 5000)

print(emp1.name)
print(emp1.salary)
```

```python
class Employee:
    def __init__(self, name, salary = 0):
        self.name = name
        if salary >= 0:
            self.salary = salary
        else:
            self.salary = 0
            print('Invalid Salary!')

emp1 = Employee('Sara', 5000)
emp2 = Employee('Ali', -100)

print(emp1.salary)
print(emp2.salary)
```
