```python
#Lab 3: Break
numbers = [1,3,5,7,9,11,13]
for n in numbers:
    if n == 7:
        break
    print (n)
print("Number 7 found");
```

```python
i = 1
while i <= 10:
    if i == 5:
        break
    print(i)
    i += 1
```

```python
#Lab 4: Continue
fruits = ['apple', 'banana', 'orange', 'pear']
for fruit in fruits:
  if fruit == 'orange':
    continue
  print(fruit)
```

```python
i = 1
while i <= 10:
  if i == 5:
    i += 1
    continue
  print(i)
  i += 1
```

```python
# Print all the numbers except for even
numbers = [1,2,3,4,5]
for n in numbers:
    if n % 2 == 0:
        continue
    print("Odd number:", n)
```
