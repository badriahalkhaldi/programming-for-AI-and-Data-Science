```python
def name_adder(names):
    new_list = []
    i = 0
    while i < len(names):
        if names[i] != "":
            new_list.append(names[i])
        i += 1
    return new_list

print(name_adder(["RAWABI", "", "ARYAM", "BADRIAH"]))
```

```python
def name_adder(names):
    new_list = []
    i = 0
    while i < len(names):
        if names[i] == "":
            print("There is an empty string")
            return new_list
        new_list.append(names[i])
        i += 1
    return new_list

print(name_adder(["RAWABI", "ARYAM", "", "BADRIAH"]))
```
