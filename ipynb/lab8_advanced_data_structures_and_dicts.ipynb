```python
import math

def convert(value, conversion_type):
    if conversion_type == "c_to_f":
        return (value * 9/5) + 32
    elif conversion_type == "f_to_c":
        return (value - 32) * 5/9
    elif conversion_type == "inch_to_cm":
        return value * 2.54
    elif conversion_type == "cm_to_inch":
        return value / 2.54
    elif conversion_type == "deg_to_rad":
        return math.radians(value)
    elif conversion_type == "rad_to_deg":
        return math.degrees(value)
    elif conversion_type == "sar_to_usd":
        return value / 3.75
    elif conversion_type == "usd_to_sar":
        return value * 3.75
    # Bonus
    elif conversion_type == "kg_to_lb":
        return value * 2.20462
    elif conversion_type == "lb_to_kg":
        return value / 2.20462
    else:
        return "Invalid conversion type"


while True:
    print("\nAvailable conversions:")
    print("1- Celsius → Fahrenheit")
    print("2- Fahrenheit → Celsius")
    print("3- Inch → Cm")
    print("4- Cm → Inch")
    print("5- Degrees → Radians")
    print("6- Radians → Degrees")
    print("7- SAR → USD")
    print("8- USD → SAR")
    print("9- Kg → Lb")
    print("10- Lb → Kg")
    print("Type 'exit' to quit")

    user_choice = input("\nChoose conversion: ")

    if user_choice == "exit":
        break

    value = float(input("Enter value: "))

    types = {
        "1": "c_to_f",
        "2": "f_to_c",
        "3": "inch_to_cm",
        "4": "cm_to_inch",
        "5": "deg_to_rad",
        "6": "rad_to_deg",
        "7": "sar_to_usd",
        "8": "usd_to_sar",
        "9": "kg_to_lb",
        "10": "lb_to_kg"
    }

    ctype = types.get(user_choice)
    print("Result:", convert(value, ctype))
```

```python
def grade_calculator(marks):
    total = sum(marks)
    average = total / len(marks)

    if average >= 90:
        grade = "A"
    elif average >= 80:
        grade = "B"
    elif average >= 70:
        grade = "C"
    elif average >= 60:
        grade = "D"
    else:
        grade = "F"

    return total, average, grade

marks_list = [80, 90, 100, 70]
total, avg, grade = grade_calculator(marks_list)
print("Total:", total)
print("Average:", avg)
print("Grade:", grade)
```

```python
def add_book():
    title = input("Enter book title: ")
    author = input("Enter author: ")
    with open("library.txt", "a") as file:
        file.write(f"{title} - {author}\n")
    print("Book added successfully.\n")


def display_books():
    try:
        with open("library.txt", "r") as file:
            print("\n--- Library Books ---")
            print(file.read())
    except FileNotFoundError:
        print("No books found.\n")


def search_book():
    title = input("Enter title to search: ").lower()
    found = False

    try:
        with open("library.txt", "r") as file:
            for line in file:
                if title in line.lower():
                    print("Found:", line)
                    found = True
    except FileNotFoundError:
        print("Library is empty.")

    if not found:
        print("Book not found.\n")


while True:
    print("\n1- Add Book")
    print("2- Display Books")
    print("3- Search Book")
    print("4- Exit")

    choice = input("Choose: ")

    if choice == "1":
        add_book()
    elif choice == "2":
        display_books()
    elif choice == "3":
        search_book()
    elif choice == "4":
        break
    else:
        print("Invalid option.")
```
