```python
# Tuples and Functions (*args)
def sum_and_multiply(*args):
    print(args)
    print(type(args))

sum_and_multiply(2, 5, 4, 3)
```

```python
# divmod example
result = divmod(23, 4)
print(result)
```

```python
# tuple unpacking
quotient, remainder = divmod(23, 4)

print('quotient:', quotient)
print('remainder:', remainder)
```

```python
# zip function
text = 'xyz'
a_list = [1, 2, 3, 4]

zip_obj = zip(text, a_list)

print(zip_obj)

for z in zip_obj:
    print(z)
```

```python
# tuple and list (list of tuples)
months_days = [
    ('January', 31),
    ('February', 28),
    ('March', 31),
    ('April', 30)
]

print(months_days)
```

```python
# function + sorted
def myFunc(e):
    return len(e)

cars = ('Mercedes', 'Audi', 'BMW', 'Porsche', 'VW')

print(myFunc('Audi'))

sorted_cars = sorted(cars, key=myFunc)
print(sorted_cars)
```

```python
# lambda function
cars = ('Mercedes', 'Audi', 'BMW', 'Porsche', 'VW')

sorted_cars = sorted(cars, key=lambda e: len(e))
print(sorted_cars)
```

```python
# check understanding
def create_tuple():
    t = tuple(range(1, 11))

    new_list = list(t)
    for i in range(11, 21):
        new_list.append(i)

    return tuple(new_list)

print(create_tuple())
```
