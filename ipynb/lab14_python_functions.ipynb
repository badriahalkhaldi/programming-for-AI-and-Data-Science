### Lab 1: Functions

```python
#Function with no arguments
def greet():
    print("Hello")

greet()
```

```python
#Function with 1 argument
def greet(name):
    print("Hello", name)

greet("Badriah")
greet(123)
```

```python
#Function with 2 arguments
def greet(name, id):
    print("Hello", name, ",ID: ", id)

greet("Badriah", 12345)
```

```python
def function_name(arguments):
    """This is a sample docstring for demo."""
    print("Inside the function.")

print(function_name.__doc__)
function_name(5)
```

```python
#User - defined function
def greet (name):
    """
    This function greets a person with their name
    """
    print("Hello, " + name + ". Good Morning!")

greet("Badriah")
```

```python
# Built-in function
#print() is built-in function
print("Hello, World!") #String
print(5) #Int
print(5.586968696) #Float
print(True) #Boolean

#len() is built-in function
list = [1,2,3]
print(len(list))
```

```python
#Positional Argument: Values must be passed in order
def greet(name, msg):
    """This function greets to
    the person with the provided message"""
    print("Hello", name + ', ' + msg)

greet("Badriah", "Good morning!")
#greet(“Badriah”) ## Error: 1 positional argument is missing
```

```python
#Default aurgument : Value that a function auotmatically uses when you don't provide one

def test(a=8, b=9):
    return a+b

#Case-1: No arguments
print(test())
#Case-2: One arguments
print(test(5))
#Case-3: Both arguments
print(test(5, 10))
```

```python
#Positional & Default aurgument

def test(a, b=9):
    return a+b

#Case-1: One arguments
print(test(5))

#Case-2: Both arguments
print(test(5, 10))
```

```python
#Keywords Arguments: allows to pass values using parameter names

def test2(**kwargs):
    return kwargs

print(test2(a=5, b=6, c=7))
#kwargs collects it into dictionary.
```

```python
def my_function (*a, **k):
    print("Arbitary arguments:", a)
    print("Keyword arguments:", k)

my_function(1,2,3,name="Badriah", age=23)
```

```python
def add(*args):
    return sum(args)

list1 = [1,2,3]
list2 = [4,5,6]
add(*list1, *list2)
#OR
add(*[1,2,3], *[4,5,6])
```

```python
def add(**kwargs):
    return sum(kwargs.values()) #sum all the values in the dictionary

add(a=2,b=3,c=5,d=10)
```

```python
#Global Variables
c=5 #Global

def add(a,b):
    return a+b+c

print(add(4,6))
print(c)
```

```python
#Local Variables
def add(a,b):
    d = 20 #local
    return a+b+d
print(add(4,11))
# print(d): gives error
```

```python
#Retrun multiple values as tuple
def calc(a,b):
    return a+b, a-b, a*b

s,d,p = calc((2,4), 1) #tuple unpacking
print("Sum: ", s)
print("Difference: ", d)
print("Product: ", p)
```

```python
count = 10
def show_count():
    count = 5
    print("Local: ",count)
print("Global: ", count)
show_count()
```

```python
count = 10 #global
def show_count():
    global count #using the global variable
    count = count + 5
    print("Inside function: ",count)
show_count()
print(count)
```

```python
count = 10 #global
def show_count(n):
    return n+5
count = show_count(count)
print(count)
```
