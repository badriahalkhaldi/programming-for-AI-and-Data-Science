```python
# Lab 1 : Dictionary
# With list
student = ["Sarah", 16, "A"]
print(student[0])

# With dictionary
student = {"name": "Sarah", "age":16, "grade": "A"}
print(student["name"]) 
```

```python
#Print the dictionary
dic = {'a' : 'ABC', 'c' : 'XYZ'}
print(dic) 
```

```python
#Duplicate Keys - Last value is kept
ph ={'aman':123,'anu':345,'iop':345,'aman':555}
print(ph)
```

```python
#Access value using key 
dic = {'a' : 'ABC', 'c' : 'XYZ'}
d1=dic['c'] 
print(d1) 
```

```python
#Add new key-value pair
dic['b']= 'QWE'
print(dic) 
```

```python
# How may key-value pairs in the dictionary?
print(len(dic)) 
```

```python
#Delete key & its value
del dic['c'] 
print(dic)
```

```python
#Change the value of a key 
dic['a']='nd' 
print(dic)
```

```python
#Deleting a dictionary
del dic  
# print(dic) will now give error as dic no longer exists
```

```python
#Check existence of a key
dic = {'a' : 'ABC','c' : 'XYZ'}
d1='a' in dic 
print(d1)
d2='a' not in dic 
print(d2)
```

```python
dic = {10 : 'ABC', 20 : 'XYZ', 30 : 'QWE'}
print(dic)
```

```python
#Return all the keys
d2=dic.keys()
print(d2)
```

```python
#Return all the values
d2=dic.values() 
print(d2)
```

```python
#Returns the value for that key
d3=dic.get('10') 
print(d3)
```

```python
#Safer- doesn't crash the program or gives KeyError
d3=dic.get('40', 'Not found') #returns not found if key doesnt exists
print(d3)
d4 = dic.get('100') #returns none by default
print(d4)
```

```python
# Add all key-value pairs from dictionary b into dictionary a
a={'a':11,'b':22,"c":33} 
b={'s':123}
a.update(b)
print(a)
```

```python
#Removes a key & returns the value
dic = {'10' : 'ABC','20' : 'XYZ','30' : 'QWE'}
d4=dic.pop('10')
print(d4)
print(dic)
```

```python
#Lab 2:Basic Statistical Operations

#Mean: The average of all numbers in a list
import statistics
print(statistics.mean([-4,-6,-8,-11]))
print(statistics.mean([1,2,3,4,5,6,7,8,9])) 
print(statistics.mean([9,12,45,23]))
```

```python
#Median is the middle value when all numbers are arranged in order
import statistics
print(statistics.median([-4,-6,-8,-11]))
print(statistics.median([1,2,3,4,5,6,7,8,9])) 
print(statistics.median([9,12,45,23]))
```

```python
#Mode is the number that appears most frequently in the list
import statistics
print(statistics.mode([-4,-6,-6,-8,-11]))
print(statistics.mode([1,2,3,4,4,5,5,7,7,7]))
print(statistics.mode(['green','blue','red','green']))
```

```python
#Standard Deviation is the distance of numbers from the mean (sqaure root of variance)
import statistics
print(statistics.stdev([-4,-6,-6,-8,-11]))
print(statistics.stdev([1,2,3,4,5,6,7,8,9])) 
print(statistics.stdev([9,12,45,23]))
```

```python
#Variance is the a measure of how spread out the numbers are from the mean
import statistics
print(statistics.variance([-4,-6,-6,-8,-11]))
print(statistics.variance([1,2,3,4,5,6,7,8,9])) 
print(statistics.variance([9,12,45,23]))
```
