## Lab 1: Data Filtering

```python
import pandas as pd
# Create a sample DataFrame
df = pd.DataFrame({'A': [1, 2, 3, 4], 'B': ['X', 'Y', 'Z', 'W']})
df
```

### Boolean Indexing

```python
#Boolean Indexing - Only the rows where the condition in TRUE.
#This works on row column 
# Filter rows where values in column 'A' are greater than 2
filtered_df = df[df['A'] > 2]
print(filtered_df)
```

### Query Method

```python
#Using Query Method just like SQL
filtered_df = df.query('A > 2')
print(filtered_df)
```

```python
filtered_df = df.query('A > 2 and B =="Z"')
print(filtered_df)
```

### Using loc and iloc

```python
# loc - label based (column names)
# Select rows where column 'A' is greater than 2
# df.loc[row_condition, column_selection]
filtered_df = df.loc[df['A'] > 2, ['A','B']]
print(filtered_df)
```

```python
# iloc - index based (row numbers)
# Show the first two rows
filtered_df = df.iloc[0:2]
print(filtered_df)
```

### Multiple Conditions

```python
# Multiple Conditions - AND &, OR|, NOT~
# Filter rows where column 'A' is greater than 2 and column 'B' equals 'Z'
filtered_df = df[(df['A'] > 2) & ~(df['B'] == 'Z')]
print(filtered_df)
```
