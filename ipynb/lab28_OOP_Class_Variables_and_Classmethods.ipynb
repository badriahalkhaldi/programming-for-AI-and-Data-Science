```python
class Employee:
    MIN_SALARY = 10000

    def __init__(self, name, salary):
        self.name = name
        if salary >= Employee.MIN_SALARY:
            self.salary = salary
        else:
            self.salary = Employee.MIN_SALARY


# إنشاء الموظفين
emp1 = Employee('John', 30000)
emp2 = Employee('Jane', 9000)

# طباعة الرواتب
print(emp1.salary)
print(emp2.salary)
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    @classmethod
    def from_file(cls, filename):
        with open(filename, 'r') as f:
            name = f.readline().strip()
            salary = int(f.readline().strip())
        return cls(name, salary)


emp1 = Employee('John', 50000)
emp2 = Employee.from_file('employee_data.txt')

print(emp2.name)
```

```python
class Person:
    CURRENT_YEAR = 2026

    def __init__(self, name, age):
        self.name = name
        self.age = age

    @classmethod
    def from_birth_year(cls, name, birth_year):
        age = cls.CURRENT_YEAR - birth_year
        return cls(name, age)

# إنشاء الكائن (هذا كان ناقص بالصورة)
bob = Person.from_birth_year("Bob", 1990)

print(bob.name)
print(bob.age)
```

```python
class BetterDate:
    def __init__(self, year, month, day):
        self.year = year
        self.month = month
        self.day = day

    @classmethod
    def from_str(cls, datestr):
        parts = datestr.split('-')
        year = int(parts[0])
        month = int(parts[1])
        day = int(parts[2])
        return cls(year, month, day)

xmas = BetterDate.from_str("2026-12-25")

print(xmas.year)
print(xmas.month)
print(xmas.day)
```
